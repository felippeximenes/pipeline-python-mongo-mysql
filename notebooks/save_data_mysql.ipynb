## Configurando o MySQL com Python

Neste momento, vamos configurar a integração entre o MySQL e Python. Isso nos permitirá executar consultas, inserções e outras operações no banco de dados MySQL usando o Python como nossa linguagem de programação principal.

```python
import mysql.connector # Importando a biblioteca

# Criando o objeto de conexao
cnx = mysql.connector.connect (
    host="localhost",
    user="FELIPPEXIMENES",
    password="12345"
)

print(cnx)

```

Criaremos um cursor, que é um objeto utilizado para executar instruções SQL no contexto do Python. Essse cursor nos permitirá enviar consultas e comandos SQL para o banco de dados MySQL por meio da conexão estabelecida e obter os resultados de volta para serem processados em nosso código Python

```python
cursor = cnx.cursor()
```

## Criando uma base de dados

O método `execute()` é usado para compilar uma instrução SQL

```python
#execute() considerado metodo que executs os comandos SQL

cursor.execute("CREATE DATABASE IF NOT EXISTS dbprodutos;") #criando a base de dados
```

## Criando uma tabela

Antes de criar nossa tabela, vamos revisar as colunas necessárias que precisaremos incluir nela. Essa etapa é importante para garantir que nossa tabela seja projetada corretamente, atendendo aos requisitos de armazenamento e organização dos dados.

```python
import pandas as pd

df_livros = pd.read_csv("/root/pipeline-python-mongo-mysql/data/tabela_livros.csv")
df_livros.head()

# Fazendo a leitura do arquivo e armazenando dentro do DATAFRAME df_livros
```

```python
df_livros.columns

# Leitura da tabela em colunas
```

```python
df_livros.shape

# Quantidade das colunas 
```

```python
# Criando a tabela no MySQL
# Lembrando que VARCHAR significa que e uma string
# FLOAT significa que sao decimais
# INT e valor inteiro
# Lembrando que nao e uma boa pratica utlizar caracteres especiais e nem espacos em uma tabela MySQL
# Precisa especificar qual destas colunas vai ser a Primary Key (chave primaria)
# A coluna de chave primaria vai possuir os valores que sao de identificador unico para cada um dos registros desta tabela
# A coluna de chave primaria sera a de ID para que todos os registros desta tabela nao poderam ter nenhuma valor repetido na coluna ID


cursor.execute("""
    CREATE TABLE IF NOT EXISTS dbprodutos.tb_livros(
               id VARCHAR(100),
               Produto VARCHAR(100),
               Categoria_Produto VARCHAR(100),
               Preco FLOAT(10,2),
               Frete FLOAT(10,2),
               Data_Compra DATE,
               Vendedor VARCHAR(100),
               Local_Compra VARCHAR(100),
               Avaliacao_Compra INT,
               Tipo_Pagamento VARCHAR(100),
               Qntd_Parcelas INT,
               Latitude FLOAT(10,2),
               Longitude FLOAT(10,2),

               PRIMARY KEY (id)
   );
                           
""")
```

**Selecionando a base de dados para verificar a tabela criada**

```python
# Verificando se a tabela doi criada 

cursor.execute("USE dbprodutos;")
cursor.execute("SHOW TABLES;")

# FOR serve para percorrer o resultado do comando SHOW TABLE

for tb in cursor:
    print(tb)
```

## Inserindo os dados do csv na tabela

Para inserir os dados na tabela do MySQL, é necessário percorrer cada linha do DataFrame e transformá-las em tuplas. Essa abordagem permite que mapeemos os dados do DataFrame para as colunas correspondentes da tabela do MySQL de forma eficiente e precisa.

```python
# Aqui sera utilizado o metodo itrrows para integrar os dados de cada linha do DATAFRAME
# Como ele e um iterador sera utlizado o comando FOR para visualizar os resultados retornados 

for i, row in df_livros.iterrows():
    print(tuple(row))

# A variavel i vai retornar cada um dos indices do DATAFRAME
# for i, row in df_livros.iterrows():
   # print(i)

# Ja a row ira retornar cada um dos dados de cada linha que poddui no DATAFRAME
# for i, row in df_livros.iterrows():
    # print (row)
    
# Os dados retornam na estrutura de PANDAS 
# No tuple retorna cada um dos dados de cada uma das linhas 
```

```python
# List comprehension permite criar listas em uma forma concisa em uma unica lista de codigos
# Nome da lista lista_dados para a criancao executando o for direto desta lista utlizando o tuple

lista_dados = [tuple(row) for i, row in df_livros.iterrows()]
lista_dados

# Dados ja estruturados dentro de uma lista dentro da tabela 
```

```python
cursor = cnx.cursor()
```

```python
import uuid # O uuid4 gera um ID aleatorio caso seja necessrio, e o erro de id persista

lista_dados = []
for i, row in df_livros.iterrows():
    novo_id = uuid.uuid4().hex  # gera um ID único # hex transforma esse UUID (que normalmente tem hífens) em uma string sem hífens, pronta pra banco de dados.
    dados = (novo_id,) + tuple(row[1:])  # substitui o valor original da coluna id
    lista_dados.append(dados)
# A lista_dados agora contém os dados com o novo ID
```

```python
# Criando a variavel e colocando o comando para inserir os dados
# Os %s representam o valores que iram inserir de uma forma dinamica, ja que sao muitos valores
# Considerado um marcados geral considerado em qualquer tabela

sql = "INSERT INTO dbprodutos.tb_livros VALUES (%s,%s,%s,%s,%s,%s,%s,%s,%s,%s,%s,%s,%s);"

cursor.executemany(sql, lista_dados)
cnx.commit() # Aqui garante que as alteracoes com este comando seja aplicada definitivamente na tabela

# Executemanny para que o comando seja executado varias vezes ja que possui muitos dados para que seja adicionado
# cnx e o objeto de conexao
```

```python
# Metodo rowcount retorna a quantidade de linhas que foram adicionadas

print(cursor.rowcount, "dados inseridos")
```

```python
# Agora vamos criar a tabela de produtos 2021 em diante
# A tabela de produtos 2021 em diante sera criada com os mesmos dados da tabela de livros

cursor.execute("""
    CREATE TABLE IF NOT EXISTS dbprodutos.tb_produtos_2021_em_diante(
               id VARCHAR(100),
               Produto VARCHAR(100),
               Categoria_Produto VARCHAR(100),
               Preco FLOAT(10,2),
               Frete FLOAT(10,2),
               Data_Compra DATE,
               Vendedor VARCHAR(100),
               Local_Compra VARCHAR(100),
               Avaliacao_Compra INT,
               Tipo_Pagamento VARCHAR(100),
               Qntd_Parcelas INT,
               Latitude FLOAT(10,2),
               Longitude FLOAT(10,2),

               PRIMARY KEY (id));
""")
```

```python
# Tabela cirada, realizando a leitura do arquivo CSV (tabela_2021_em_diante.csv)
# Conferindo a quantidade de linhas e colunas
# Lembrando que o arquivo CSV esta na pasta data


import pandas as pd
df_produtos = pd.read_csv("../data/tabela_2021_em_diante.csv")
df_produtos.head()
```

```python
# Conferindo a quantidade de linhas e colunas
df_produtos.shape 
```

```python
import uuid #basicamente o uuid4 gera um ID aleatorio caso seja necessrio, e o erro de id persista

lista_dados = []
for i, row in df_produtos.iterrows():
    novo_id = uuid.uuid4().hex
    dados = (
        novo_id,
        row['Produto'],
        row['Categoria do Produto'],
        row['Preço'],
        row['Frete'],
        row['Data da Compra'],
        row['Vendedor'],
        row['Local da compra'],
        row['Avaliação da compra'],
        row['Tipo de pagamento'],
        row['Quantidade de parcelas'],
        row['Latitude'],
        row['Longitude']
    )
    lista_dados.append(dados)

sql = "INSERT INTO dbprodutos.tb_produtos_2021_em_diante VALUES (%s,%s,%s,%s,%s,%s,%s,%s,%s,%s,%s,%s,%s)"
cursor.executemany(sql, lista_dados)
cnx.commit()
```
